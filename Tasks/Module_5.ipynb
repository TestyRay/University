### **Содержимое модуля transformations.py:**

```python
%%writefile geo_transform/transformations.py

import math


class CartesianCoordinates:
    """
    Representation of Cartesian coordinates.
    """

    def __init__(self, x: float, y: float, z: float):
        self.x = x
        self.y = y
        self.z = z

    def __repr__(self) -> str:
        return f"CartesianCoordinates(x={self.x}, y={self.y}, z={self.z})"


class SphericalCoordinates:
    """
    Representation of spherical coordinates.
    """

    def __init__(self, r: float, theta: float, phi: float):
        self.r = r
        self.theta = theta
        self.phi = phi

    def __repr__(self) -> str:
        return (
            f"SphericalCoordinates("
            f"r={self.r}, "
            f"theta={self.theta}, "
            f"phi={self.phi})"
        )


class CoordinateConverter:
    """
    Converter between Cartesian and spherical coordinate systems.
    """

    @staticmethod
    def to_spherical(cart: CartesianCoordinates) -> SphericalCoordinates:
        """
        Convert CartesianCoordinates to SphericalCoordinates.
        """
        r = (
            math.sqrt(cart.x**2 + cart.y**2 + cart.z**2)
        )
        theta = (
            2 * math.atan2(cart.y, cart.x + math.sqrt(cart.x**2 + cart.y**2))
        )
        phi = (
            math.atan2(math.sqrt(cart.x**2 + cart.y**2), cart.z)
        )

        return SphericalCoordinates(r, theta, phi)

    @staticmethod
    def to_cartesian(sph: SphericalCoordinates) -> CartesianCoordinates:
        """
        Convert SphericalCoordinates to CartesianCoordinates.
        """
        x = sph.r * math.sin(sph.phi) * math.cos(sph.theta)
        y = sph.r * math.sin(sph.phi) * math.sin(sph.theta)
        z = sph.r * math.cos(sph.phi)

        return CartesianCoordinates(x, y, z)
```

### **Содержимое модуля utils.py:**

```python
%%writefile geo_transform/utils.py

import math


class AngleConverter:
    """
    Convert angles between degrees and radians.
    """

    @staticmethod
    def deg_to_rad(degrees: float) -> float:
        return degrees * math.pi / 180

    @staticmethod
    def rad_to_deg(radians: float) -> float:
        return radians * 180 / math.pi
```

### **Содержимое модуля file_operations.py:**

```python
%%writefile geo_transform/file_operations.py

from .transformations import CartesianCoordinates


class FileManager:
    """
    File operations: reading coordinates and writing results.
    """

    @staticmethod
    def write_text(filename: str, text: str) -> None:
        with open(filename, "w", encoding="utf-8") as file:
            file.write(text)

    @staticmethod
    def read_cartesian(filename: str) -> list[CartesianCoordinates]:
        result = []

        with open(filename, "r", encoding="utf-8") as file:
            for line in file:
                parts = line.strip().split()
                if len(parts) == 3:
                    x, y, z = map(float, parts)
                    result.append(CartesianCoordinates(x, y, z))

        return result
```

```python
%%writefile geo_transform/__main__.py

from .transformations import (
    CartesianCoordinates,
    SphericalCoordinates,
    CoordinateConverter,
)
from .utils import (
    AngleConverter,
)
from .file_operations import (
    FileManager,
)


def main():
    print("Меню работы с пакетом geo_transform")
    print("1 — Преобразование декартовых координат в сферические")
    print("2 — Преобразование сферических координат в декартовые")
    print("3 — Перевод градусов в радианы")
    print("4 — Перевод радиан в градусы")
    print("5 — Чтение координат из файла и преобразование")
    print("6 — Запись текста в файл")

    choice = input("Выберите пункт меню: ")

    if choice == "1":
        x = float(input("Введите x: "))
        y = float(input("Введите y: "))
        z = float(input("Введите z: "))

        cart = CartesianCoordinates(x, y, z)
        sph = CoordinateConverter.to_spherical(cart)

        print("Результат преобразования:")
        print(sph)

    elif choice == "2":
        r = float(input("Введите r: "))
        th_deg = float(input("Введите угол theta (в градусах): "))
        ph_deg = float(input("Введите угол phi (в градусах): "))

        theta = AngleConverter.deg_to_rad(th_deg)
        phi = AngleConverter.deg_to_rad(ph_deg)

        sph = SphericalCoordinates(r, theta, phi)
        cart = CoordinateConverter.to_cartesian(sph)

        print("Результат преобразования:")
        print(cart)

    elif choice == "3":
        deg = float(input("Введите угол в градусах: "))
        print("Значение в радианах:", AngleConverter.deg_to_rad(deg))

    elif choice == "4":
        rad = float(input("Введите угол в радианах: "))
        print("Значение в градусах:", AngleConverter.rad_to_deg(rad))

    elif choice == "5":
        filename = input("Введите имя файла: ")
        coords = FileManager.read_cartesian(filename)

        if not coords:
            print("Файл пуст или содержит ошибки.")
            return

        print("Результат преобразования координат:")
        for c in coords:
            sph = CoordinateConverter.to_spherical(c)
            print(c, "→", sph)

    elif choice == "6":
        filename = input("Введите имя файла: ")
        text = input("Введите текст: ")
        FileManager.write_text(filename, text)
        print("Текст записан.")

    else:
        print("Неизвестный пункт меню.")


if __name__ == "__main__":
    main()
```

### **Содержимое модуля main.py (с импортом пакета и тестированием функций из него):**

```python
from geo_transform import (
    CartesianCoordinates,
    SphericalCoordinates,
    CoordinateConverter,
    FileManager,
)
from geo_transform.utils import (
    AngleConverter,
)


def print_title(title: str) -> None:
    print("\n" + "=" * 60)
    print(title)
    print("=" * 60)


def test_cartesian_to_spherical():
    print_title("Тест 1. Преобразование декартовых координат в сферические")

    cart = CartesianCoordinates(3, 4, 5)
    sph = CoordinateConverter.to_spherical(cart)

    print(f"Исходные координаты: x = {cart.x}, y = {cart.y}, z = {cart.z}")
    print("Результат преобразования:")
    print(f"Радиус (r): {sph.r}")
    print(f"theta (радианы): {sph.theta}")
    print(f"phi (радианы): {sph.phi}")


def test_spherical_to_cartesian():
    print_title("Тест 2. Преобразование сферических координат в декартовые")

    r = 10
    theta_deg = 45
    phi_deg = 60

    theta = AngleConverter.deg_to_rad(theta_deg)
    phi = AngleConverter.deg_to_rad(phi_deg)

    sph = SphericalCoordinates(r, theta, phi)
    cart = CoordinateConverter.to_cartesian(sph)

    print("Исходные значения:")
    print(f"Радиус: {r}")
    print(f"theta угол: {theta_deg} градусов")
    print(f"phi угол: {phi_deg} градусов")
    print("Результат преобразования:")
    print(f"x = {cart.x}")
    print(f"y = {cart.y}")
    print(f"z = {cart.z}")


def test_angle_conversion():
    print_title("Тест 3. Перевод углов между градусами и радианами")

    degrees = 180
    radians = AngleConverter.deg_to_rad(degrees)
    back_to_degrees = AngleConverter.rad_to_deg(radians)

    print(f"{degrees} градусов в радианах: {radians}")
    print(f"{radians} радиан обратно в градусы: {back_to_degrees}")


def test_file_operations():
    print_title("Тест 4. Запись и чтение данных из файла")

    filename = "test_coords.txt"
    data = "1 2 3\n4 5 6\n10 20 30\n"

    print(f"Создаём файл '{filename}' и записываем в него координаты...")
    FileManager.write_text(filename, data)

    print("Читаем данные из файла:")
    coords = FileManager.read_cartesian(filename)
    print(coords)

    print("\nПреобразуем прочитанные координаты в сферические:")

    for point in coords:
        sph = CoordinateConverter.to_spherical(point)
        print(f"Точка ({point.x}, {point.y}, {point.z}) преобразована в:")
        print(f"r = {sph.r}")
        print(f"theta = {sph.theta}")
        print(f"phi = {sph.phi}")
        print()


def main():
    print_title("Демонстрация работы пакета geo_transform")

    test_cartesian_to_spherical()
    test_spherical_to_cartesian()
    test_angle_conversion()
    test_file_operations()

    print("\nВсе тесты выполнены.")


if __name__ == "__main__":
    main()
```
