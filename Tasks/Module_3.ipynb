## **Задание №1. Создайте класс Sphere для представления сферы в трехмерном пространстве**


Обеспечьте следующие методы класса:

1. Конструктор, принимающий 4 действительных числа: радиус, и 3 координаты центра шара. Если конструктор вызывается без аргументов, создать объект сферы с единичным радиусом и центром в начале координат. Если конструктор вызывается с аргументом, создать объект сферы с соответствующим радиусом и центром в начале координат.

2. Метод get_volume (), который возвращает действительное число — объем шара, ограниченной текущей сферой.

3. Метод get_square (), который возвращает действительное число — площадь внешней поверхности сферы.

4. Метод get_radius (), который возвращает действительное число — радиус сферы.

5. Метод get_center (), который возвращает тьюпл с 3 действительными числами — координатами центра сферы в том же порядке, в каком они задаются в конструкторе.

6. Метод set_radius (r), который принимает 1 аргумент — действительное число, и меняет радиус текущей сферы, ничего не возвращая.

7. Метод set_center (x, y, z), который принимает 3 аргумента — действительных числа, и меняет координаты центра сферы, ничего не возвращая. Координаты задаются в том же порядке, что и в конструкторе.

8. Метод is_point_inside (x, y, z), который принимает 3 аргумента — действительных числа — координаты некоторой точки в пространстве (в том же порядке, что и в конструкторе), и возвращает логическое значение True или False в зависимости от того, находится эта точка внутри сферы.

````python
import math

from typing import (
    Union,
)
````

````python
class Sphere:
    """
    Represent sphere in 3D space.
    """
    def __init__(
        self,
        radius: Union[int, float] = 1.0,
        x: Union[int, float] = 0.0,
        y: Union[int, float] = 0.0,
        z: Union[int, float] = 0.0,
    ) -> None:
        """
        Initialize sphere with radius and center coordinates.

        :param radius: Sphere radius.
        :param x: X coordinate of center.
        :param y: Y coordinate of center.
        :param z: Z coordinate of center.
        """
        self.radius = float(radius)
        self.x = float(x)
        self.y = float(y)
        self.z = float(z)

    def get_volume(self) -> float:
        """
        Return sphere volume.

        :return: Volume of sphere.

        Example:
            >>> Sphere(0.5).get_volume()
            0.5235987755982988
        """
        return 4 / 3 * math.pi * self.radius ** 3

    def get_square(self) -> float:
        """
        Return surface area of sphere.

        :return: Sphere surface area.

        Example:
            >>> Sphere(0.5).get_square()
            6.283185307179586
        """
        return 4 * math.pi * self.radius ** 2

    def get_radius(self) -> float:
        """
        Return sphere radius.

        :return: Sphere radius.

        Example:
            >>> Sphere(5).get_radius()
            5.0
        """
        return self.radius

    def get_center(self) -> tuple[float, float, float]:
        """
        Return center coordinates of sphere.

        :return: Tuple of (x, y, z).

        Example:
            >>> Sphere(0.5).get_center()
            (0.0, 0.0, 0.0)
        """
        return (self.x, self.y, self.z)

    def set_radius(self, radius: Union[int, float]) -> None:
        """
        Set new radius for sphere.

        :param radius: New radius.

        Example:
            >>> sphere = Sphere(1)
            >>> sphere.get_radius()
            1.0
            >>> sphere.set_radius(0.5)
            >>> sphere.get_radius()
            0.5
        """
        self.radius = float(radius)

    def set_center(
        self,
        x: Union[int, float],
        y: Union[int, float],
        z: Union[int, float],
    ) -> None:
        """
        Set new center coordinates of sphere.

        :param x: New X coordinate.
        :param y: New Y coordinate.
        :param z: New Z coordinate.

        Example:
            >>> s = Sphere(0.5)
            >>> s.set_center(1, 2, 3)
            >>> s.get_center()
            (1.0, 2.0, 3.0)
        """
        self.x = float(x)
        self.y = float(y)
        self.z = float(z)

    def is_point_inside(self, x: float, y: float, z: float) -> bool:
        """
        Check if point (x,y,z) is inside sphere.

        :param x: X coordinate of point.
        :param y: Y coordinate of point.
        :param z: Z coordinate of point.

        :return: True if point inside or on surface, False otherwise.

        Example:
            >>> Sphere(1.6).is_point_inside(0 , -1.5, 0)
            True
        """
        distance = math.sqrt(
            (x - self.x) ** 2 +
            (y - self.y) ** 2 +
            (z - self.z) ** 2
        )

        return distance <= self.radius
````

````python
# Tests
s0 = Sphere(0.5)
assert s0.get_center() == (0.0, 0.0, 0.0)

assert math.isclose(s0.get_volume(), 0.5235987755982988)

assert s0.is_point_inside(0, -1.5, 0) is False

s0.set_radius(1.6)

assert s0.get_radius() == 1.6

assert s0.is_point_inside(0, -1.5, 0) is True

print("All tests passed")
````

## **Задание №2. Создайте класс SuperStr, который наследует стандартный класс str и содержит 2 новых метода:**

1. Метод is_repeatance (s), который принимает 1 аргумент s и возвращает True или False в зависимости от того, может ли текущая строку быть получена целым количеством повторов строки s. Вернуть False, если s не является строкой. Считать, что пустая строка не содержит повторов.

2. Метод is_palindrom (), который возвращает True или False в зависимости от того, является ли строка [палиндромом](https://ru.wikipedia.org/wiki/Палиндром). Регистрами символов пренебрегать. Пустую строку считать палиндромом.

````python
class SuperStr(str):
    """
    Extended string class with additional methods.
    """

    def is_repeatance(self, pattern: str) -> bool:
        """
        Return True if current string is made of repeating pattern.

        :param pattern: String pattern to check repetition.

        :return: True if current string equals pattern repeated several times,
                 else False.

        Example:
            >>> SuperStr("abcabcabc").is_repeatance("abc")
            True
            >>> SuperStr("ababccab").is_repeatance("abc")
            False
            >>> SuperStr("abc").is_repeatance(123)
            False
        """
        if not pattern or not isinstance(pattern, str):
            return False

        return not(any(self.split(pattern)))

    def is_palindrom(self) -> bool:
        """
        Check if current string is palindrome (case-insensitive).

        :return: True if string is palindrome, otherwise False.

        Example:
            >>> SuperStr("level").is_palindrom()
            True
            >>> SuperStr("LeVel").is_palindrom()
            True
            >>> SuperStr("test").is_palindrom()
            False
        """
        text = self.lower()
        return text == text[::-1]
````

````python
# Tests
s = SuperStr("123123123123")

assert s.is_repeatance("123") is True
assert s.is_repeatance("123123") is True
assert s.is_repeatance("123123123123") is True
assert s.is_repeatance("12312") is False
assert s.is_repeatance(123) is False
assert s.is_palindrom() is False
assert str(s) == "123123123123"
assert int(s) == 123123123123
assert s + "qwe" == "123123123123qwe"

p = SuperStr("123_321")
assert p.is_palindrom() is True

print("All tests passed")
````

````python
class Student:
    """
    Represent student with name, surname, gender, courses and grades.
    """
    def __init__(self, name: str, surname: str, gender: str):
        """
        Initialize student basic information.

        :param name: Student name.
        :param surname: Student surname.
        :param gender: Student gender.
        """
        self.name: str = name
        self.surname: str = surname
        self.gender: str = gender
        self.finished_courses: list[str] = []
        self.courses_in_progress: list[str] = []
        self.grades: dict[str, list[int]] = {}

    def add_courses(self, course_name: str) -> None:
        """
        Add finished course to student record.

        :param course_name: Course name to add.
        """
        self.finished_courses.append(course_name)

    def rate_lecturer(
        self,
        lecturer: "Lecturer",
        course: str,
        grade: int,
    ) -> None:
        """
        Add grade to lecturer if both are attached to same course.

        :param lecturer: Lecturer object to rate.
        :param course: Course name.
        :param grade: Integer grade (1–10).

        Example:
            >>> student.rate_lecturer(lecturer, "Python", 9)
        """
        if not isinstance(lecturer, Lecturer):
            raise ValueError(
                "Lecturer must be instance of Lecturer class."
                )

        if (
            course not in self.courses_in_progress
            or course not in lecturer.courses_attached
        ):
            raise ValueError(
                "Lecturer and student must be attached to same course."
            )

        lecturer.grades.setdefault(course, []).append(grade)

    def _get_average_grade(self) -> float:
        """
        Return average grade for student's homework.
        """
        all_grades = [
            grade
            for grades in self.grades.values()
            for grade in grades
        ]

        return (
            round(sum(all_grades) / len(all_grades), 1)
            if all_grades else
            0.0
        )

    def __str__(self) -> str:
        """
        Return formatted student info.
        """
        in_progress = ", ".join(self.courses_in_progress) or "нет"
        finished = ", ".join(self.finished_courses) or "нет"
        return (
            f"Имя: {self.name}\n"
            f"Фамилия: {self.surname}\n"
            f"Средняя оценка за домашние задания: {self._get_average_grade()}\n"
            f"Курсы в процессе изучения: {in_progress}\n"
            f"Завершенные курсы: {finished}"
        )

    def __str__(self) -> str:
        """
        Return formatted student info.

        :return: String with student info
        """
        in_progress = ", ".join(self.courses_in_progress) or "нет"
        finished = ", ".join(self.finished_courses) or "нет"

        return (
            f"Имя: {self.name}\n"
            f"Фамилия: {self.surname}\n"
            f"Средняя оценка за домашние задания: {self._get_average_grade()}\n"
            f"Курсы в процессе изучения: {in_progress}\n"
            f"Завершенные курсы: {finished}"
        )

    def __eq__(self, other: "Student") -> bool:
        """
        Compare two students by average grade.

        :param other: Another student

        :return: True if average grades are equal.
        """
        if not isinstance(other, Student):
            return NotImplemented

        return self._get_average_grade() == other._get_average_grade()

    def __lt__(self, other: "Student") -> bool:
        """
        Compare if this student has lower average grade.

        :param other: Another student.

        :return: True if self < other.
        """
        if not isinstance(other, Student):
            return NotImplemented

        return self._get_average_grade() < other._get_average_grade()

    def __len__(self) -> int:
        """
        Return number of finished courses.

        :return: Count of finished courses.
        """
        return len(self.finished_courses)

class Lecturer(Mentor):
    """
    Represent lecturer who receives grades from students.
    """

    def __init__(self, name: str, surname: str):
        """
        Initialize lecturer and add grades dictionary.
        """
        super().__init__(name, surname)
        self.grades: dict[str, list[int]] = {}

    def _get_average_grade(self) -> float:
        """
        Return average grade for lectures.
        """
        all_grades = [
            grade
            for grades in self.grades.values()
            for grade in grades
        ]

        return (
            round(sum(all_grades) / len(all_grades), 1)
            if all_grades else
            0.0
        )

    def __str__(self) -> str:
        """
        Return formatted lecturer info.
        """
        return (
            f"Имя: {self.name}\n"
            f"Фамилия: {self.surname}\n"
            f"Средняя оценка за лекции: {self._get_average_grade()}"
        )

    def __eq__(self, other: "Lecturer") -> bool:
        """
        Compare lecturers by average grade.

        :param other: Another lecturer

        :return: True if average grades are equal.
        """
        if not isinstance(other, Lecturer):
            return NotImplemented

        return self._get_average_grade() == other._get_average_grade()

    def __lt__(self, other: "Lecturer") -> bool:
        """
        Compare lecturers by average grade (less than).

        :param other: Another lecturer

        :return: True if self < other.
        """
        if not isinstance(other, Lecturer):
            return NotImplemented

        return self._get_average_grade() < other._get_average_grade()

    def __add__(self, other: "Lecturer") -> "Lecturer":
        """
        Combine grades of two lecturers.

        :param other: Lecturer to combine with

        :return: New Lecturer with merged grades.
        """
        if not isinstance(other, Lecturer):
            return NotImplemented

        new_lecturer = Lecturer(
            f"{self.name}&{other.name}",
            f"{self.surname}&{other.surname}"
        )
        new_lecturer.grades = {
            course: self.grades.get(course, []) + other.grades.get(course, [])
            for course in set(self.grades) | set(other.grades)
        }
        return new_lecturer

class Reviewer(Mentor):
    """
    Represent reviewer who rates student homework.
    """

    def rate_hw(self, student: Student, course: str, grade: int) -> None:
        """
        Add grade to student for specific course if allowed.

        :param student: Student object to rate
        :param course: Course name
        :param grade: Integer grade (1–10)

        Example:
            >>> reviewer.rate_hw(student, "Python", 8)
        """
        if not isinstance(student, Student):
            raise ValueError(
                "Student must be instance of Student class."
            )

        if (
            course not in self.courses_attached
            or course not in student.courses_in_progress
        ):
            raise ValueError(
                "Reviewer and student must be attached to same course."
            )

        student.grades.setdefault(course, []).append(grade)

    def __str__(self) -> str:
        """
        Return formatted reviewer info.
        """
        return f"Имя: {self.name}\nФамилия: {self.surname}"

    def __len__(self) -> int:
        """
        Return number of attached courses.

        :return: Count of attached courses.
        """
        return len(self.courses_attached)

    def __eq__(self, other: "Reviewer") -> bool:
        """
        Compare two reviewers by name and surname.

        :param other: Another reviewer

        :return: True if name and surname are equal.
        """
        if not isinstance(other, Reviewer):
            return NotImplemented

        return (self.name, self.surname) == (other.name, other.surname)

    def __repr__(self) -> str:
        """
        Return short debug representation for reviewer.

        :return: String like Reviewer(Sergey Petrov).
        """
        return f"Reviewer({self.name} {self.surname})"
````

4. **Дополнительные методы класса:**

   - **`reciprocal()`**: возвращает новую дробь, обратную данной (числитель и знаменатель меняются местами).
   - **Строковое представление** (`__str__`): возвращает строку вида `"числитель/знаменатель"`. Если знаменатель равен 1, возвращает только числитель в виде строки.
   - **Официальное представление** (`__repr__`): возвращает строку вида `"Fraction(числитель, знаменатель)"`.

5. **Обработка исключений:**

   - При попытке создать дробь с нулевым знаменателем выводить в консоль сообщение `ValueError`.
   - При попытке деления на нулевую дробь, выводить в консоль сообщение `ZeroDivisionError`.

6. **Тестирование:**

   - Написать код, демонстрирующий работу всех реализованных методов и операторов, используя примеры, приведенные ниже.
   - Обеспечить корректную обработку исключений и вывод соответствующих сообщений.





````python
from fractions import (
    Fraction as PyFraction,
)
from math import (
    gcd,
)


class Fraction:
    """
    Represent mathematical fraction with numerator and denominator.
    """

    def __init__(self, numerator: int, denominator: int):
        """
        Initialize fraction with numerator and denominator.
        Automatically simplify the fraction.

        :param numerator: Fraction numerator
        :param denominator: Fraction denominator
        :raise ValueError: If denominator equals zero
        """
        if denominator == 0:
            raise ValueError("denominator cannot be zero")

        if denominator < 0:
            numerator, denominator = -numerator, -denominator

        self.numerator: int = numerator
        self.denominator: int = denominator
        self._simplify()

    def _simplify(self) -> None:
        """
        Simplify fraction to its lowest terms.
        """
        divisor = gcd(self.numerator, self.denominator)
        self.numerator //= divisor
        self.denominator //= divisor

    def __add__(self, other: "Fraction") -> "Fraction":
        """
        Return sum of two fractions as new Fraction.

        :param other: Another Fraction.

        :return: New simplified Fraction.
        """
        if not isinstance(other, Fraction):
            return NotImplemented

        num = (
            self.numerator * other.denominator
            + other.numerator * self.denominator
        )
        den = self.denominator * other.denominator

        return Fraction(num, den)

    def __sub__(self, other: "Fraction") -> "Fraction":
        """
        Return difference of two fractions as new Fraction.

        :param other: Another Fraction.

        :return: New simplified Fraction.
        """
        if not isinstance(other, Fraction):
            return NotImplemented

        num = (
            self.numerator * other.denominator
            - other.numerator * self.denominator
        )
        den = self.denominator * other.denominator

        return Fraction(num, den)

    def __mul__(self, other: "Fraction") -> "Fraction":
        """
        Return product of two fractions as new Fraction.

        :param other: Another Fraction.

        :return: New simplified Fraction.
        """
        if not isinstance(other, Fraction):
            return NotImplemented

        num = self.numerator * other.numerator
        den = self.denominator * other.denominator

        return Fraction(num, den)

    def __truediv__(self, other: "Fraction") -> "Fraction":
        """
        Return division result of two fractions as new Fraction.

        :param other: Another Fraction.

        :return: New simplified Fraction.

        :raise ZeroDivisionError: If other fraction is zero.
        """
        if not isinstance(other, Fraction):
            return NotImplemented

        if other.numerator == 0:
            raise ZeroDivisionError(
                "Division by a fraction of zero is not possible"
            )

        num = self.numerator * other.denominator
        den = self.denominator * other.numerator

        return Fraction(num, den)

    def __eq__(self, other: "Fraction") -> bool:
        """
        Compare equality of two fractions.

        :param other: Another Fraction.

        :return: True if fractions are equal.
        """
        return (
            isinstance(other, Fraction)
            and (
                self.numerator * other.denominator
                == self.denominator * other.numerator
            )
        )

    def __lt__(self, other: "Fraction") -> bool:
        """
        Check if this fraction is less than other.

        :param other: Another Fraction.

        :return: True if less than.
        """
        return (
            self.numerator * other.denominator
            < other.numerator * self.denominator
        )

    def __le__(self, other: "Fraction") -> bool:
        """
        Check if this fraction is less or equal to other.

        :param other: Another Fraction.

        :return: True if less or equal.
        """
        return (
            self.numerator * other.denominator
            <= other.numerator * self.denominator
        )

    def __gt__(self, other: "Fraction") -> bool:
        """
        Check if this fraction is greater than other.

        :param other: Another Fraction.

        :return: True if greater than.
        """
        return (
            self.numerator * other.denominator
            > other.numerator * self.denominator
        )

    def __ge__(self, other: "Fraction") -> bool:
        """
        Check if this fraction is greater or equal to other.

        :param other: Another Fraction.

        :return: True if greater or equal.
        """
        return (
            self.numerator * other.denominator
            >= other.numerator * self.denominator
        )

    def __float__(self) -> float:
        """
        Convert fraction to float.

        :return: Fraction value as float.
        """
        return self.numerator / self.denominator

    def reciprocal(self) -> "Fraction":
        """
        Return new fraction reciprocal to this one.

        :return: New Fraction with swapped numerator and denominator.

        :raise ZeroDivisionError: If numerator equals zero.
        """
        if self.numerator == 0:
            raise ZeroDivisionError("Zero fraction has no inverse")

        return Fraction(self.denominator, self.numerator)

    @classmethod
    def from_float(cls, value: float) -> "Fraction":
        """
        Create Fraction from float value.

        :param value: Float number.

        :return: Fraction representing given float.
        """
        f = PyFraction(value).limit_denominator()

        return cls(f.numerator, f.denominator)

    def __str__(self) -> str:
        """
        Return user-friendly string representation.

        :return: String like '3/4' or '3' if denominator == 1
        """
        return (
            str(self.numerator)
            if self.denominator == 1 else
            f"{self.numerator}/{self.denominator}"
        )

    def __repr__(self) -> str:
        """
        Return official representation of fraction.

        :return: String like 'Fraction(3, 4)'.
        """
        return f"Fraction({self.numerator}, {self.denominator})"
````
