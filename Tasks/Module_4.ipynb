## **Теоретическая часть:**

1. **Преобразование декартовых координат в сферические:**

   *Формулы для преобразования:*

   $
   r = \sqrt{x^2 + y^2 + z^2}
   $

   - Эта формула вычисляет радиус $r$ от начала координат до точки с координатами $(x, y, z)$ в декартовой системе. Это расстояние в пространстве.

   $
   \theta = 2 \cdot \text{atan2}(y, x + \sqrt{x^2 + y^2})
   $

   - Формула для $\theta$ (азимутальный угол) использует функцию $\text{atan2}$, которая возвращает угол в радианах между положительной осью $x$ и точкой $(x, y)$. Умножение на 2 и добавление $\sqrt{x^2 + y^2}$ в знаменатель обеспечивает корректное определение угла в диапазоне от 0 до $2\pi$.

   $
   \phi = \text{atan2}(\sqrt{x^2 + y^2}, z)
   $

   - Формула для $\phi$ (полярный угол) также использует $\text{atan2}$, но здесь она вычисляет угол между положительной осью $z$ и вектором от начала координат до точки. Это значение также получается в радианах.

   **Значения $\theta$ и $\phi$ получаются в радианах. Для перевода их в градусы необходимо умножить на $\frac{180}{\pi}$.**





2. **Преобразование сферических координат в декартовые:**

  *Формулы для преобразования:*

   $
   x = r \cdot \sin(\phi) \cdot \cos(\theta)
   $
   
   - Эта формула вычисляет координату $x$ в декартовой системе, используя радиус $r$, полярный угол $\phi$ и азимутальный угол $\theta$.

   $
   y = r \cdot \sin(\phi) \cdot \sin(\theta)
   $

   - Аналогично, эта формула вычисляет координату $y$ в декартовой системе.

   $
   z = r \cdot \cos(\phi)
   $
   
   - Эта формула вычисляет координату $z$ в декартовой системе, используя радиус $r$ и полярный угол $\phi$.

   **Значения $\theta$ и $\phi$ должны быть заданы в радианах. Если они заданы в градусах, их необходимо предварительно перевести в радианы, умножив на $\frac{\pi}{180}$.**

3. **Преобразование между градусами и радианами:**

   $
   \text{градусы} \rightarrow \text{радианы}: \text{degrees} \cdot \frac{\pi}{180}
   $
   - Эта формула используется для перевода углов из градусов в радианы.

   $
   \text{радианы} \rightarrow \text{градусы}: \text{radians} \cdot \frac{180}{\pi}
   $
   - Эта формула используется для перевода углов из радиан в градусы.

```python
import math

# Пример расчета синуса и косинуса от угла в градусах
angle_degrees = 45  # Угол в градусах

# Преобразование градусов в радианы
angle_radians = math.radians(angle_degrees)

# Расчет синуса и косинуса
sin_value = math.sin(angle_radians)
cos_value = math.cos(angle_radians)

print(f"Синус {angle_degrees} градусов: {sin_value}")
print(f"Косинус {angle_degrees} градусов: {cos_value}")
```

```python
# Пример расчета atan2 и преобразования результата в градусы
y = 1
x = 5

# Расчет atan2 в радианах
atan2_radians = math.atan2(y, x)

# Преобразование результата atan2 в градусы
atan2_degrees = math.degrees(atan2_radians)

print(f"atan2({y}, {x}) в радианах: {atan2_radians}")
print(f"atan2({y}, {x}) в градусах: {atan2_degrees}")
```

### **Содержимое модуля transformations.py:**

```python
%%writefile geo_transform/transformations.py

import math


def cartesian_to_spherical(
    x: float,
    y: float,
    z: float,
) -> tuple[float, float, float]:
    """
    Convert Cartesian coordinates to spherical coordinates.

    :param x: X coordinate
    :param y: Y coordinate
    :param z: Z coordinate

    :return: Tuple (r, theta, phi)
    """
    r = math.sqrt(x ** 2 + y ** 2 + z ** 2)
    theta = 2 * math.atan2(y, x + math.sqrt(x ** 2 + y ** 2))
    phi = math.atan2(math.sqrt(x ** 2 + y ** 2), z)
    return r, theta, phi


def spherical_to_cartesian(
    r: float,
    theta: float,
    phi: float,
) -> tuple[float, float, float]:
    """
    Convert spherical coordinates to Cartesian coordinates.

    :param r: Radius value
    :param theta: Azimuth angle in radians
    :param phi: Polar angle in radians

    :return: Tuple (x, y, z)
    """
    x = r * math.sin(phi) * math.cos(theta)
    y = r * math.sin(phi) * math.sin(theta)
    z = r * math.cos(phi)
    return x, y, z
```

### **Содержимое модуля utils.py:**

```python
%%writefile geo_transform/utils.py

import math


def deg_to_rad(degrees: float) -> float:
    """
    Convert degrees to radians.

    :param degrees: Angle in degrees

    :return: Angle in radians
    """
    return degrees * math.pi / 180


def rad_to_deg(radians: float) -> float:
    """
    Convert radians to degrees.

    :param radians: Angle in radians

    :return: Angle in degrees
    """
    return radians * 180 / math.pi
```

### **Содержимое модуля file_operations.py:**

```python
%%writefile geo_transform/file_operations.py

def write_results_to_file(filename: str, data: str) -> None:
    """
    Write text data to file.

    :param filename: File path
    :param data: Text content
    """
    with open(filename, "w", encoding="utf-8") as file:
        file.write(data)


def read_coordinates_from_file(
    filename: str,
) -> list[tuple[float, float, float]]:
    """
    Read coordinates from file.

    :param filename: File path

    :return: List tuples (x, y, z)
    """
    result = []
    with open(filename, "r", encoding="utf-8") as file:
        for line in file:
            parts = line.strip().split()
            if len(parts) == 3:
                x, y, z = map(float, parts)
                result.append((x, y, z))
    return result
```

```python
%%writefile geo_transform/__main__.py

from .transformations import cartesian_to_spherical, spherical_to_cartesian
from .utils import deg_to_rad, rad_to_deg
from .file_operations import write_results_to_file, read_coordinates_from_file


def main():
    print("Меню работы с пакетом geo_transform")
    print("1 — Преобразование декартовых координат в сферические")
    print("2 — Преобразование сферических координат в декартовые")
    print("3 — Перевод угла из градусов в радианы")
    print("4 — Перевод угла из радиан в градусы")
    print("5 — Чтение координат из файла и преобразование в сферические")
    print("6 — Запись текста в файл")

    choice = input("Выберите пункт меню: ")

    if choice == "1":
        x = float(input("Введите координату x: "))
        y = float(input("Введите координату y: "))
        z = float(input("Введите координату z: "))

        r, theta, phi = cartesian_to_spherical(x, y, z)

        print("Результат преобразования:")
        print(f"Радиус r: {r}")
        print(f"Азимутальный угол theta (в радианах): {theta}")
        print(f"Полярный угол phi (в радианах): {phi}")

    elif choice == "2":
        r = float(
            input("Введите радиус r: ")
        )
        theta_deg = float(
            input("Введите азимутальный угол theta (в градусах): ")
        )
        phi_deg = float(
            input("Введите полярный угол phi (в градусах): ")
        )

        theta = deg_to_rad(theta_deg)
        phi = deg_to_rad(phi_deg)

        x, y, z = spherical_to_cartesian(r, theta, phi)

        print("Результат преобразования:")
        print(f"Координата x: {x}")
        print(f"Координата y: {y}")
        print(f"Координата z: {z}")

    elif choice == "3":
        degrees = float(input("Введите угол в градусах: "))
        radians = deg_to_rad(degrees)
        print(f"Значение угла в радианах: {radians}")

    elif choice == "4":
        radians = float(input("Введите угол в радианах: "))
        degrees = rad_to_deg(radians)
        print(f"Значение угла в градусах: {degrees}")

    elif choice == "5":
        filename = input("Введите имя файла: ")

        coords = read_coordinates_from_file(filename)
        if not coords:
            print("Файл пуст или содержит неверные данные.")
            return

        print("Результаты преобразования координат из файла:")
        for x, y, z in coords:
            r, theta, phi = cartesian_to_spherical(x, y, z)
            print(f"Точка ({x}, {y}, {z}) преобразована в:")
            print(f"r = {r}")
            print(f"theta = {theta}")
            print(f"phi = {phi}")
            print()

    elif choice == "6":
        filename = input("Введите имя файла: ")
        text = input("Введите текст для записи: ")

        write_results_to_file(filename, text)
        print(f"Текст успешно записан в файл: {filename}")

    else:
        print("Введён неизвестный пункт меню.")


if __name__ == "__main__":
    main()
```

```python
from geo_transform import (
    cartesian_to_spherical,
    spherical_to_cartesian,
    deg_to_rad,
    rad_to_deg,
    write_results_to_file,
    read_coordinates_from_file,
)


def print_title(title: str) -> None:
    print("\n" + "=" * 60)
    print(title)
    print("=" * 60)


def test_cartesian_to_spherical():
    print_title("Тест 1. Преобразование декартовых координат в сферические")

    x, y, z = 3, 4, 5
    r, theta, phi = cartesian_to_spherical(x, y, z)

    print(f"Исходные координаты: x = {x}, y = {y}, z = {z}")
    print("Результат преобразования:")
    print(f"Радиус (r): {r}")
    print(f"theta (радианы): {theta}")
    print(f"phi (радианы): {phi}")


def test_spherical_to_cartesian():
    print_title("Тест 2. Преобразование сферических координат в декартовые")

    r = 10
    theta_deg = 45
    phi_deg = 60

    theta = deg_to_rad(theta_deg)
    phi = deg_to_rad(phi_deg)

    x, y, z = spherical_to_cartesian(r, theta, phi)

    print(f"Исходные значения:")
    print(f"Радиус: {r}")
    print(f"theta угол: {theta_deg} градусов")
    print(f"phi угол: {phi_deg} градусов")
    print("Результат преобразования:")
    print(f"x = {x}")
    print(f"y = {y}")
    print(f"z = {z}")


def test_angle_conversions():
    print_title("Тест 3. Перевод углов между градусами и радианами")

    degrees = 180
    radians = deg_to_rad(degrees)
    back_to_degrees = rad_to_deg(radians)

    print(f"{degrees} градусов в радианах: {radians}")
    print(f"{radians} радиан обратно в градусы: {back_to_degrees}")


def test_file_operations():
    print_title("Тест 4. Запись и чтение данных из файла")

    filename = "test_coords.txt"
    data = "1 2 3\n4 5 6\n10 20 30\n"

    print(f"Создаём файл '{filename}' и записываем в него координаты...")
    write_results_to_file(filename, data)

    print("Читаем данные из файла:")
    coords = read_coordinates_from_file(filename)
    print(coords)

    print("\nПреобразуем прочитанные координаты в сферические:")
    for x, y, z in coords:
        r, theta, phi = cartesian_to_spherical(x, y, z)
        print(f"Точка ({x}, {y}, {z}) преобразована в:")
        print(f"r = {r}")
        print(f"theta = {theta}")
        print(f"phi = {phi}")
        print()


def main():
    print_title("Демонстрация работы пакета geo_transform")

    test_cartesian_to_spherical()
    test_spherical_to_cartesian()
    test_angle_conversions()
    test_file_operations()

    print("\nВсе тесты выполнены.")


if __name__ == "__main__":
    main()
```
